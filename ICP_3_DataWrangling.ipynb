```python
df = pd.read_csv("Sacramentorealestatetransactions.csv")
df.head()
```

Number of columns in Data frame

```python
df.shape[1]
```

Mean of house prices

```python
mean_price = df['price'].mean()
print(mean_price)
```

Median of the house prices

```python
median_price = df['price'].median()
print(median_price)
```

Different types of property in the dataset

```python
property_types = sorted(df['type'].dropna().unique())
print(property_types)
```

Maximum price of a "condo" in the dataset

```python
max_condo_price = df.loc[df["type"].str.lower() == "condo", "price"].max()
print(max_condo_price)
```

Mean of each type of property in the dataset

```python
mean_prices = df.groupby(df["type"].str.lower())["price"].mean()
print(mean_prices)
```

Count of different property types in dataset

```python
type_counts = df["type"].str.lower().value_counts()
print(type_counts)
```
